```python
def count_words(paragraph):
    paragraph = paragraph.lower()
    # Removing basic punctuation
    paragraph = paragraph.replace(".", "")
    paragraph = paragraph.replace(",", "")
    paragraph = paragraph.replace("!", "")
    paragraph = paragraph.replace("?", "")
    paragraph = paragraph.replace(":", "")
    paragraph = paragraph.replace(";", "")
    paragraph = paragraph.replace('"', "")
    paragraph = paragraph.replace("'", "")

    words=paragraph.split()
    frequency={}

    for word in words:
        if word in frequency:
            frequency[word]+=1
        else:
            frequency[word]=1
    return frequency

paragraph = input("Enter a paragraph: ")
print(paragraph)
frequency = count_words(paragraph)

print("\nWord Frequency:")

for word in frequency:
    print(word, ":", frequency[word])

# Finding the highest frequency
highest = 0

for word in frequency:
    if frequency[word] > highest:
        highest = frequency[word]

print("\nMost Frequently Occurring Word(s):")

for word in frequency:
    if frequency[word] == highest:
        print(word, ":", highest, "times")

print("\nWords Occurring Only Once:")

for word in frequency:
    if frequency[word] == 1:
        print(word)
```
