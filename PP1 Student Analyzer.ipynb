```python
def calculate_marks(marks):
    total=sum(marks)
    percentage=total/5
    return total,percentage
```

```python
n=int(input("Enter Number of Students:"))
subjects=("Algebra","Mechanics","Calculas","Chemistry","Electromagnetism")
students={}
for student in range(n):
    name=input("\nEnter Student's Name:")
    marks=[]
    for sub in subjects:
        mark=int(input("Enter marks in "+sub+":"))
        marks.append(mark)

    students[name]=marks
```

```python
for name in students:
    marks=students[name]
    total, percentage=calculate_marks(marks)

    if percentage>=90:
        grade="A"
    elif percentage>=75:
        grade="B"
    elif percentage>=60:
        grade="C"
    elif percentage>=40:
        grade="D"
    else:
        grade="F"

    if percentage>=40:
        status="PASS !"
    else:
        status="FAIL !"

    print("\nName of student:",name)
    print("Marks:",marks)
    print("Total:",total)
    print("Percentage:",percentage)
    print("Grade:",grade)
    print("Status:",status)
```

```python
#highest score
highest_scorer=""
highest_total=0
for student in students:
    total=sum(students[student])
    if total>highest_total:
        highest_total=total 
        highest_scorer=student

print("Highest Scorer: ", highest_scorer)
print("Total Marks: ",highest_total)
```
