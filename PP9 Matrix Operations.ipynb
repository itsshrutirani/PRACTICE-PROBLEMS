```python
# Function to take matrix input
def input_matrix(rows, columns):

    matrix = []

    for i in range(rows):

        row = []

        for j in range(columns):
            value = int(input("Enter value: "))
            row.append(value)

        matrix.append(row)

    return matrix


# Function to display matrix
def display_matrix(matrix):

    for row in matrix:
        print(row)


# Function for matrix addition
def add_matrix(A, B):

    result = []

    for i in range(len(A)):

        row = []

        for j in range(len(A[0])):
            value = A[i][j] + B[i][j]
            row.append(value)

        result.append(row)

    return result


# Function for transpose
def transpose_matrix(A):

    result = []

    for j in range(len(A[0])):

        row = []

        for i in range(len(A)):
            row.append(A[i][j])

        result.append(row)

    return result


# Function for multiplication
def multiply_matrix(A, B):

    result = []

    for i in range(len(A)):

        row = []

        for j in range(len(B[0])):

            total = 0

            for k in range(len(B)):
                total += A[i][k] * B[k][j]

            row.append(total)

        result.append(row)

    return result


# Taking matrix size
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

print("\nEnter Matrix A:")
A = input_matrix(rows, columns)

print("\nEnter Matrix B:")
B = input_matrix(rows, columns)

# Addition
print("\nMatrix Addition:")
addition = add_matrix(A, B)
display_matrix(addition)

# Transpose
print("\nTranspose of Matrix A:")
transpose = transpose_matrix(A)
display_matrix(transpose)

# Multiplication
if columns == rows:

    print("\nMatrix Multiplication:")
    multiplication = multiply_matrix(A, B)
    display_matrix(multiplication)

else:
    print("\nMultiplication not possible.")
```
