```python
# Product catalogue
products = {
    "Laptop": {"price": 50000, "stock": 5},
    "Phone": {"price": 20000, "stock": 10},
    "Headphones": {"price": 2000, "stock": 15},
    "Mouse": {"price": 500, "stock": 20}
}

cart = {}

def show_products():
    print("\nAvailable Products:")

    for name in products:
        print(name, "- Price:", products[name]["price"],
              "- Stock:", products[name]["stock"])

def add_product():
    name = input("Enter product name: ")

    if name in products:
        quantity = int(input("Enter quantity: "))

        if quantity <= 0:
            print("Enter a valid quantity.")

        elif quantity > products[name]["stock"]:
            print("Not enough stock available.")

        else:
            if name in cart:
                cart[name] += quantity
            else:
                cart[name] = quantity

            products[name]["stock"] -= quantity
            print("Product added to cart.")

    else:
        print("Product not found.")


def remove_product():
    name = input("Enter product name to remove: ")

    if name in cart:
        quantity = int(input("Enter quantity to remove: "))

        if quantity <= 0:
            print("Enter a valid quantity.")

        elif quantity > cart[name]:
            print("You do not have that many items in your cart.")

        else:
            cart[name] -= quantity
            products[name]["stock"] += quantity

            if cart[name] == 0:
                del cart[name]

            print("Product removed.")

    else:
        print("Product is not in your cart.")


def generate_bill():
    subtotal = 0

    print("\n--- YOUR BILL ---")

    for name in cart:
        quantity = cart[name]
        price = products[name]["price"]
        amount = price * quantity

        print(name, "x", quantity, "=", amount)

        subtotal += amount

    if subtotal >= 2000:
        discount = subtotal * 0.10
    else:
        discount = 0

    final_amount = subtotal - discount

    print("Subtotal:", subtotal)
    print("Discount:", discount)
    print("Final Amount:", final_amount)


while True:
    print("\n--- SHOPPING MENU ---")
    print("1. Show Products")
    print("2. Add Product")
    print("3. Remove Product")
    print("4. Generate Bill")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        show_products()

    elif choice == 2:
        add_product()

    elif choice == 3:
        remove_product()

    elif choice == 4:
        generate_bill()

    elif choice == 5:
        print("Thank you for shopping!")
        break

    else:
        print("Invalid choice.")
```
