```python
products = {
    "Pen": {"price": 20, "quantity": 5, "category": "Stationery", "reorder": 10},
    "Notebook": {"price": 100, "quantity": 20, "category": "Stationery", "reorder": 5},
    "Mouse": {"price": 500, "quantity": 3, "category": "Electronics", "reorder": 5},
    "Keyboard": {"price": 1000, "quantity": 8, "category": "Electronics", "reorder": 3}
}


# Products below reorder level
print("Products needing reorder:")

for name in products:

    quantity = products[name]["quantity"]
    reorder = products[name]["reorder"]

    if quantity < reorder:
        print(name)


# Total inventory value
total_value = 0

for name in products:

    price = products[name]["price"]
    quantity = products[name]["quantity"]

    total_value = total_value + price * quantity


print("\nTotal Inventory Value:", total_value)


# Find most valuable category
category_value = {}

for name in products:

    category = products[name]["category"]
    value = products[name]["price"] * products[name]["quantity"]

    if category in category_value:
        category_value[category] = category_value[category] + value
    else:
        category_value[category] = value


highest_category = ""
highest_value = 0

for category in category_value:

    if category_value[category] > highest_value:
        highest_value = category_value[category]
        highest_category = category


print("Most Valuable Category:", highest_category)
print("Value:", highest_value)
```
