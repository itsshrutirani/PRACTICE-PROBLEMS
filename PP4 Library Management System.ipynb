```python
books={
    "python":True,
    "c programming":True,
    "java":True,
    "data science":True
}

def show_books():
    print("\nAvailable Books")

    for book in books:
        if books[book]==True:
            print(book)

def issue_book():
    name=input("Enter book name to issue:")

    if name in books:
       if books[name]==True:
           books[name]=False
           print("Book issued Succesfully!")

       else:
           print("Book already issued.")
    else:
        print("Book ot found.")

def return_book():
    name=input("Enter book name to return: ")
    if name in books:
        if books[name]==False:
            books[name]=True

            late_days=int(input("Enter no of late days:"))
            if late_days>0:
                fine=late_days*5
            else:
                fine=0
            print("Book returned succesfully!")
            print("Fine:",fine)

        else:
            print("This book was not issued")

def unavailable_books():
    print("\nUnavailable Books:")

    found = False

    for book in books:
        if books[book] == False:
            print(book)
            found = True

    if found == False:
        print("No books are currently issued.")


# Main menu
while True:
    print("\n--- LIBRARY MENU ---")
    print("1. Show Available Books")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Show Unavailable Books")
    print("5. Exit")

    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        show_books()

    elif choice == 2:
        issue_book()

    elif choice == 3:
        return_book()

    elif choice == 4:
        unavailable_books()

    elif choice == 5:
        print("Thank you!")
        break

    else:
        print("Invalid choice.")
```
