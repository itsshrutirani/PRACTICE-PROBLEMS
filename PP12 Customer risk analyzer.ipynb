```python
customers = {}


def analyze_customer(deposits, withdrawals):

    total_deposit = sum(deposits)
    total_withdrawal = sum(withdrawals)

    transaction_count = len(deposits) + len(withdrawals)

    if total_deposit == 0:
        risk = "High Risk"

    elif total_withdrawal > total_deposit * 0.80:
        risk = "High Risk"

    else:
        risk = "Normal"

    return total_deposit, total_withdrawal, transaction_count, risk


n = int(input("Enter number of customers: "))


for i in range(n):

    name = input("\nEnter customer name: ")

    deposits = []
    withdrawals = []

    d = int(input("How many deposits? "))

    for j in range(d):
        amount = float(input("Enter deposit amount: "))
        deposits.append(amount)


    w = int(input("How many withdrawals? "))

    for j in range(w):
        amount = float(input("Enter withdrawal amount: "))
        withdrawals.append(amount)


    total_deposit, total_withdrawal, count, risk = analyze_customer(
        deposits, withdrawals
    )


    customers[name] = {
        "deposits": total_deposit,
        "withdrawals": total_withdrawal,
        "transactions": count,
        "risk": risk
    }


print("\n--- CUSTOMER ANALYSIS ---")


for name in customers:

    print("\nCustomer:", name)
    print("Total Deposits:", customers[name]["deposits"])
    print("Total Withdrawals:", customers[name]["withdrawals"])
    print("Transaction Count:", customers[name]["transactions"])
    print("Risk:", customers[name]["risk"])
```
