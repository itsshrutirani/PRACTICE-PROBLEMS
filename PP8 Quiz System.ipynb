```python
questions = {
    1: ("What is the capital of India?", ("Mumbai", "Delhi", "Pune", "Chennai"), "Delhi"),
    2: ("Which language are we learning?", ("Python", "Java", "C++", "HTML"), "Python"),
    3: ("How many days are there in a week?", ("5", "6", "7", "8"), "7")
}


def start_quiz():

    score = 0
    wrong_answers = []

    for number in questions:

        question, options, answer = questions[number]

        print("\nQuestion", number)
        print(question)

        print("A.", options[0])
        print("B.", options[1])
        print("C.", options[2])
        print("D.", options[3])

        user_answer = input("Enter answer: ")

        if user_answer.lower() == answer.lower():
            print("Correct!")
            score = score + 1

        else:
            print("Wrong!")
            wrong_answers.append((question, answer))


    print("\n--- RESULT ---")
    print("Score:", score, "/", len(questions))

    percentage = (score / len(questions)) * 100

    if percentage >= 80:
        category = "Excellent"
    elif percentage >= 60:
        category = "Good"
    elif percentage >= 40:
        category = "Average"
    else:
        category = "Needs Improvement"

    print("Performance:", category)

    print("\nIncorrect Questions:")

    for question, answer in wrong_answers:
        print(question)
        print("Correct answer:", answer)


start_quiz()
```
