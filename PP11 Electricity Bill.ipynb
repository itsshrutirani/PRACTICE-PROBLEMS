```python
def calculate_bill(units):
    
    fixed_charge=100
    
    if units<=100:
        energy_charge=units*2
    elif units<=200:
        energy_charge=(100*2)+((units-100)*3)
    else: 
        energy_charge=(100*2)+(100*3)+((units-200)*5)
        
    subtotal= energy_charge + fixed_charge

    if units>300:
        surcharge=subtotal*0.10
    else:
        surcharge=0

    total=subtotal+surcharge
    return energy_charge, fixed_charge, surcharge, total

name=input("Enter consumer name:")
consumer_number = input("Enter consumer number: ")
units = int(input("Enter units consumed: "))

energy_charge, fixed_charge, surcharge, total = calculate_bill(units)

# Display bill
print("\n-----------------------------")
print("       ELECTRICITY BILL")
print("-----------------------------")

print("Consumer Name:", name)
print("Consumer Number:", consumer_number)
print("Units Consumed:", units)

print("Energy Charge:", energy_charge)
print("Fixed Charge:", fixed_charge)
print("Surcharge:", surcharge)
print("Total Bill:", total)

print("-----------------------------")
```
