```python
employees = {}


def calculate_salary(salary):

    if salary <= 30000:
        deduction = salary * 0.05

    elif salary <= 60000:
        deduction = salary * 0.10

    else:
        deduction = salary * 0.15

    net_salary = salary - deduction

    return deduction, net_salary


n = int(input("Enter number of employees: "))

for i in range(n):

    name = input("\nEnter employee name: ")
    salary = float(input("Enter basic salary: "))

    deduction, net_salary = calculate_salary(salary)

    employees[name] = {
        "salary": salary,
        "deduction": deduction,
        "net_salary": net_salary
    }


print("\n--- EMPLOYEE DETAILS ---")

for name in employees:

    print("\nName:", name)
    print("Gross Salary:", employees[name]["salary"])
    print("Deduction:", employees[name]["deduction"])
    print("Net Salary:", employees[name]["net_salary"])


# Find highest net salary

highest_name = ""
highest_salary = 0

for name in employees:

    salary = employees[name]["net_salary"]

    if salary > highest_salary:
        highest_salary = salary
        highest_name = name


print("\nEmployee with highest net salary:")
print(highest_name)
print("Net Salary:", highest_salary)
```
